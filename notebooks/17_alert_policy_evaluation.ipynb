# Phase 17 — Validated Alert-Threshold & Decision Policy Evaluation

This notebook evaluates decision policies on held-out validation data:
1. **Fixed Operating Thresholds** ($\tau_0 \in \{0.50, 0.60, 0.65, 0.75\}$)
2. **Traffic-Density-Conditioned Threshold**:
   $$\tau_{\text{alert}} = \tau_0 + \alpha \times \rho_{\text{density}}$$
3. **Calibrated Uncertainty-Aware Policy**:
   $$\tau_{\text{alert}} = \tau_0 + \beta \times \sigma_{\text{uncertainty}}$$

Evaluated across validation split videos from AI City Challenge Track 4.
