# Phase 8 Extended — Dynamic Interaction-Graph Edge Pruning Ablation

This experiment compares graph topology constructions on real traffic sequences:
- **G0**: Full Interaction Graph (all pairwise edges $i \neq j$)
- **G1**: $k$-Nearest-Neighbor Graph ($k=4$)
- **G2**: Distance-Gated Graph ($d \le 0.25$ in normalized image space)
- **G3**: Validated Hybrid Graph ($d \le 0.25$ OR converging relative motion $\Delta \mathbf{x} \cdot \Delta \mathbf{v} < 0$ and $TTC_{\text{proxy}} \le 4.0\text{s}$)

Evaluated on real tracking histories from AI City Challenge Track 4.
